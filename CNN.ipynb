```python
import matplotlib.pyplot as plt
import os
import cv2  # Ensure OpenCV is imported for reading and displaying images

def show_images_from_directory(directory, num_images=5):
    # List subfolders (classes)
    categories = os.listdir(directory)

    # Display a few images from each category
    for category in categories:
        category_path = os.path.join(directory, category)
        if os.path.isdir(category_path):
            print(f"Category: {category}")
            # List image files in the category folder
            images = os.listdir(category_path)
            # Select a few images
            selected_images = images[:num_images]

            # Display each image
            for image_name in selected_images:
                image_path = os.path.join(category_path, image_name)
                img = cv2.imread(image_path)
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for matplotlib
                plt.imshow(img)
                plt.title(category)
                plt.axis('off')
                plt.show()

# Set the path for your dataset directory
train_dir = r"D:\New folder (2)\Train_Augmented-20250124T062019Z-001\Train_dir"

# Show images from the training dataset
show_images_from_directory(train_dir, num_images=1)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Data augmentation and preprocessing for training
train_datagen = ImageDataGenerator(
    rescale=1./255,           # Rescale pixel values between 0 and 1
    shear_range=0.2,          # Apply shear transformation
    zoom_range=0.2,           # Randomly zoom in images
    horizontal_flip=True      # Flip images horizontally
)

# Data preprocessing for validation (no augmentation, just rescaling)
valid_datagen = ImageDataGenerator(rescale=1./255)

# Data preprocessing for test set
test_datagen = ImageDataGenerator(rescale=1./255)

# Define directories for train, validation, and test sets
train_dir = r"D:\New folder (2)\Train_Augmented-20250124T062019Z-001\Train_dir"
valid_dir = r"D:\New folder (2)\Train_Augmented-20250124T062019Z-001\Val_dir"
test_dir = r"D:\New folder (2)\Train_Augmented-20250124T062019Z-001\Test_dir"

# Load training data
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(150, 150),   # Resize images to a common size
    batch_size=32,
    class_mode='categorical'  # Assuming multiple disease categories
)

# Load validation data
valid_generator = valid_datagen.flow_from_directory(
    valid_dir,
    target_size=(150, 150),   # Correct single '=' for parameter assignment
    batch_size=32,
    class_mode='categorical'
)

# Load test data
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(150, 150),   # Correct single '=' for parameter assignment
    batch_size=32,
    class_mode='categorical'
)
```

```python
model = Sequential()

# Convolutional layer 1
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Convolutional layer 2
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Convolutional layer 3
model.add(Conv2D(128, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flattening the layers
model.add(Flatten())

# Fully connected layer 1
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))  # Dropout to reduce overfitting

# Fully connected layer 2 (Output layer)
model.add(Dense(train_generator.num_classes, activation='softmax'))
```

```python
model.compile(
    loss='categorical_crossentropy',  # Since we have multiple classes
    optimizer=Adam(learning_rate=0.001),
    metrics=['accuracy']
)
```

```python
import tensorflow as tf
tf.test.gpu_device_name()
```

```python
history = model.fit(
    train_generator,
    steps_per_epoch=len(train_generator),
    epochs=60,  # You can adjust the number of epochs
    validation_data=valid_generator,
    validation_steps=len(valid_generator)
)
```

```python
test_loss, test_accuracy = model.evaluate(test_generator)
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```
